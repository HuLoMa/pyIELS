### Imports

```python
from electron_state import ElectronState, etensor
from qutip_objects import kick, free_electron
from qutip import *
import matplotlib.pyplot as plt
from plot_utils import plot_2d_field
from electron_wigner import ewigner
from physics_utils import PhysicalQuantities
import numpy as np
```

### Note

This notebook focuses on presenting the results from figure 4. Some details on the computations can be found in figure_1.ipynb.

# Global definitions

```python
# Computed electron energy levels
n_electron = 51
# Computed photon energy levels
n_photon = 50

# Coupling constant
g0 = 1.0

# Object to define all the relevant physical quantities
phq = PhysicalQuantities(lambda_p=515, # in nm
                         high_tension=200, # in kV
                         zlp_fwhm=0.7, # in eV
                         momentum_resolution=40, # number of points per energy levels
                         unit_system='atomic')


a = tensor(destroy(n_photon), qeye(n_electron))
b = tensor(qeye(n_photon),kick(n_electron,1))
ps = g0*b*a.dag()  - np.conj(g0)*b.dag()*a
s = ps.expm()

electron = ElectronState(free_electron(n_electron,0))
```

# Fig. 4. a. and b.

These figures present the interaction between a tri-sqeezed vacuum state of light and a free electron.

```python
z = 0.2*np.exp(-1j*np.pi/4) # Squeezing parameter (complex number)

a = destroy(n_photon) # Annhiliation operator
squeezing = ((a**3*np.conj(z) - a.dag()**3*z)/3).expm()
vac = basis(n_photon, 0) # Vacuum state

# Squeezed vacuum state
tri_squeezed_photon = squeezing * vac
```

## Fig 4. a. tri-squeezed photon vacuum state wigner function

```python
plot_wigner(tri_squeezed_photon,np.linspace(-5,5,200),np.linspace(-5,5,200))
plt.axhline(0,color = 'grey', alpha = 0.2, linestyle = '--')
plt.axvline(0,color = 'grey', alpha = 0.2,linestyle = '--')
```

## Fig 4. b. Electron wigner function after interaction with a tri-squeezed photon vacuum state

```python
init_epts = etensor(tri_squeezed_photon,electron)
dm_epts = init_epts*init_epts.dag()
final_epts = s*dm_epts*s.dag()
final_ets = final_epts.ptrace([1])

w, ps,z = ewigner(final_ets,phq=phq,energy_levels=True)

plot_2d_field(w,z,ps,
              ylim = (-12,12),
              xlim = (0,2*np.pi),
              normalize = True,
              use_pi_ticks = True,
              use_enl_ticks = True,
              unit_system='',
              grid=True,
              projection = False)
```

# Fig. 4. c. and d.

These figures present the interaction between a quad-sqeezed vacuum state of light and a free electron.

```python
z = 0.1*np.exp(-1j*np.pi/4) # Squeezing parameter (complex number)

a = destroy(n_photon) # Annhiliation operator
squeezing = ((a**4*np.conj(z) - a.dag()**4*z)/4).expm()
vac = basis(n_photon, 0) # Vacuum state

# Squeezed vacuum state
quad_squeezed_photon = squeezing * vac
```

## Fig 4. c. quad-squeezed photon vacuum state wigner function

```python
plot_wigner(quad_squeezed_photon,np.linspace(-5,5,200),np.linspace(-5,5,200))
plt.axhline(0,color = 'grey', alpha = 0.2, linestyle = '--')
plt.axvline(0,color = 'grey', alpha = 0.2,linestyle = '--')
```

## Fig 4. d. Electron wigner function after interaction with a tri-squeezed photon vacuum state

```python
init_epqs = etensor(quad_squeezed_photon,electron)
dm_epqs = init_epqs*init_epqs.dag()
final_epqs = s*dm_epqs*s.dag()
final_eqs = final_epqs.ptrace([1])

w, ps,z = ewigner(final_eqs,phq=phq,energy_levels=True)

plot_2d_field(w,z,ps,
              ylim = (-12,12),
              xlim = (0,2*np.pi),
              normalize = True,
              use_pi_ticks = True,
              use_enl_ticks = True,
              unit_system='',
              grid=True,
              projection = False)
```
